# XML RPC client sandbox
Sandbox for developing code to develop XML RPC calls  
https://docs.python.org/3/library/xmlrpc.client.html

```python
import xmlrpc.client
import random
```

```python
with xmlrpc.client.ServerProxy("http://192.168.56.10:8001/") as proxy:
    print("3 is even: %s" % str(proxy.is_even(3)))
    print("100 is even: %s" % str(proxy.is_even(100)))
```

```python
a = ['a', 'b', 'c']
```

```python
chosen = None
```

```python
candidate = random.choice(a)
```

```python
while candidate is chosen:
    print("IN LOOP: " + candidate)
    candidate = random.choice(a)
```

```python
chosen = candidate
print("Chosen: " + chosen)
```
