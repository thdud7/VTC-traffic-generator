# VTC Automation Sandbox
Workspace to investigate playwright automation for driving VTC applications and integrating VTC bots. 
https://playwright.dev/python/

```python
import asyncio
from playwright.async_api import async_playwright
```

```python
async def main():
    async with async_playwright() as p:
        browser = await p.chromium.launch(args=["--use-fake-ui-for-media-stream", "--use-fake-device-for-media-stream"], headless=False)
        page = await browser.new_page(ignore_https_errors=True)
        await page.goto("https://192.168.56.1:8443/automated_vtc")
        #await page.goto("https://localhost:8443/")
        print(await page.title())
        await asyncio.sleep(30)
        #await browser.close()
```

```python
# Proper call for main()
# asyncio.run(main())

# Because executing in jupyter notebook event loop:
await main()
```
